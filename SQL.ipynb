```python
import pandas as pd
import pymysql
from getpass import getpass

# Database details
host = "18.136.157.135"
port = 3306
database = "project_hr_analytics"
username = "dm_team9"

password = getpass("Enter MySQL password: ")

# Connect to MySQL
connection = pymysql.connect(
    host=host,
    port=port,
    user=username,
    password=password,
    database=database
)

print("Connected successfully!")

# Check total rows
count_query = "SELECT COUNT(*) AS total_rows FROM HR"

count_df = pd.read_sql(count_query, connection)

print(count_df)
```

```python
query = "SELECT * FROM HR"

df = pd.read_sql(query, connection)

print("Rows:", len(df))
print("Columns:", len(df.columns))

df.head()
```

```python
df.to_csv("HR_complete.csv", index=False)

print("CSV created successfully!")
```

```python
df.to_excel("HR_complete.xlsx", index=False)

print("Excel file created successfully!")
```
